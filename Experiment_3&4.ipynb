# **Experiment 3.** Perform Exploratory Data Analysis for a given data set by creating Scatter Plot, Pair Plot, Count Plot using Matplotlib and Seaborn

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Column names
columns = [
    'SepalLength',
    'SepalWidth',
    'PetalLength',
    'PetalWidth',
    'Species'
]

# Load dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

df = pd.read_csv(url, names=columns)

# Display dataset
print("First Five Records")
print(df.head())

print("\nDataset Shape")
print(df.shape)

print("\nDataset Information")
print(df.info())

print("\nStatistical Summary")
print(df.describe())

print("\nMissing Values")
print(df.isnull().sum())
```

```python
# Scatter Plot
plt.figure(figsize=(4,3))
sns.scatterplot(
    data=df,
    x='SepalLength',
    y='PetalLength',
    hue='Species',
    s=80
)
plt.title("Scatter Plot")
plt.show()
```

```python
# Pair Plot
sns.pairplot(df, hue='Species')
plt.show()
```

```python
# Count Plot
plt.figure(figsize=(7,5))
sns.countplot(data=df, x='Species')

plt.title("Count Plot of Species")
plt.xlabel("Species")
plt.ylabel("Count")

plt.show()
```

# **Experiment 4.** Create Heat Maps, WordCloud

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# PART A : Heat Map
columns = [
    'SepalLength',
    'SepalWidth',
    'PetalLength',
    'PetalWidth',
    'Species'
]

url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"
iris = pd.read_csv(url, names=columns)
print("Correlation Matrix")
iris.info()
```

```python
corr = iris.iloc[:,0:4].corr()
print(corr)

plt.figure(figsize=(8,6))
sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)
plt.title("Correlation Heat Map")
plt.show()
```

```python
# PART B : Word Cloud
sms = pd.read_csv(
    "SMSSpamCollection",
    sep='\t',
    names=['Label','Message']
)
print("\nFirst Five Messages")
print(sms.head())

text = " ".join(sms["Message"])

wordcloud = WordCloud(
    width=1000,
    height=500,
    background_color='white'
).generate(text)

plt.figure(figsize=(12,6))
plt.imshow(wordcloud)
plt.axis("off")
plt.title("Word Cloud")
plt.show()
```
