```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.naive_bayes import GaussianNB

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
# Load Dataset
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data"

columns = [
'ID','Diagnosis',
'Radius','Texture','Perimeter','Area','Smoothness',
'Compactness','Concavity','ConcavePoints','Symmetry',
'FractalDimension','Radius_SE','Texture_SE','Perimeter_SE',
'Area_SE','Smoothness_SE','Compactness_SE','Concavity_SE',
'ConcavePoints_SE','Symmetry_SE','FractalDimension_SE',
'Radius_W','Texture_W','Perimeter_W','Area_W',
'Smoothness_W','Compactness_W','Concavity_W',
'ConcavePoints_W','Symmetry_W','FractalDimension_W'
]
```

```python
df = pd.read_csv(url, names=columns)
print("First Five Records")
print(df.head())
```

```python
# Feature Selection
X = df[['Radius',
        'Texture',
        'Perimeter',
        'Area',
        'Smoothness',
        'Compactness']]
```

```python
# Target Encoding
y = df['Diagnosis'].map({'M':1, 'B':0})

# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Feature Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# Gaussian Naive Bayes Model
model = GaussianNB()
model.fit(X_train, y_train)
# Prediction
y_pred = model.predict(X_test)
```

```python
# Class Probabilities
probability = model.predict_proba(X_test)

print("\nFirst 10 Predicted Class Probabilities\n")
print(probability[:10])
```

```python
# Performance Metrics
print("\nAccuracy :", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

```python
# Predict on New Samples
sample = [[14.2, 20.5, 92.4, 650.3, 0.095, 0.120]]
sample = scaler.transform(sample)
prediction = model.predict(sample)
prob = model.predict_proba(sample)
print("\nPrediction for New Sample :", prediction[0])
print("Prediction Probability :", prob)
```
